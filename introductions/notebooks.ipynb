````python
# hay algo que ayuda mucho a la hora de programar y son los prints
# ejemplo: dada una lista simple sumar 1 a cada entrada

x = [5,1,2,6,3]
for i in x:
    print("i =", i)
    ii = i + 1
    print("i + 1=", ii)

# al ser para visualizar no hace falta hacerlo perfecto, solo observar que lo estamos haciendo como bien
````

# About notebooks

Como tenemos bloques o celdas de código que pueden ejecutarse independientemente, podemos usar esto para separar responsabilidades como, definir en distintos bloques las `librerias` que vamos a usar, los `datasets` que vamos a cargar, el `analisis exploratorio`, los `modelos` que vamos a hacer y de ser necesario la `visualizacion` en plots de nuestros resultados

Esta forma de trabajo suena bastante estricta en un principio *(a mi me parecia asi y de rebelde hacia lo opuesto)* 

Sin embargo, ayuda mucho cuando dejas una tarea a medio hacer y vuelves con nuevas ideas porque se te hace facil leer tu trabajo sin perderte y abrumarte 

*Aprendí a ser ordenado despues de muchos fallos como para poder decir que a alguien como yo le ha servido pero cada quien es distinto*

```bash
# flujo de trabajo
.
├── datasets
│   └── raw/ # aqui datos iniciales o crudos
│   └── processed/ # una vez procesado recomiendo separarlo y dejarlo aquí
├── notebooks # espacio o carpeta experimental para prueba y error
│   └── 01.ipynb # puede ser 1 o 20 dependiendo que quieras probar
...
```

## How write a notebook

Bueno, como bien decia, este formato de archivo te permite separar en bloques distintos de codigos responsabilidades distintas pero el orden es importante, pensemos en esto como una cascada de trabajo y para ilustrar:

1. **Definir nuestro enfoque de modelo: Descriptivo, Explicativo, Predictivo**
2. **Sincronizar librerias o asegurar el tenerlas descargadas** 
3. **Importar librerias**
4. **Cargar el dataset**
5. **Declarar nuestras funciones a usar para ahorrar tiempo**
6. **Hacer `EDA`**
7. **Aplicar nuestro enfoque: Descriptivo, Explicativo, Predictivo**
8. **Evaluar resultados**
9. **Conclusiones**

no siempre tiene que tener esta forma, es solo como recomendacion
